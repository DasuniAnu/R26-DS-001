```python
# CELL 5 — TEST FULL FAKE GENERATION (FIRST 5 VIDEOS ONLY)

import os
import subprocess
import random

out_dir = "/content/drive/MyDrive/deepfake_v2/fake_full_v2"
os.makedirs(out_dir, exist_ok=True)

random.seed(42)

done = 0
failed = []

for v in videos:

    vid = os.path.basename(v).replace(".mp4", "")
    donor = random.choice(donors)
    output = f"{out_dir}/{vid}_fake.mp4"

    if os.path.exists(output) and os.path.getsize(output) > 500000:
        print("Already done:", vid)
        done += 1
        continue

    print("\nProcessing:", vid)

    result = subprocess.run(
        [
            "python", "roop/run.py",
            "-s", donor,
            "-t", v,
            "-o", output,
            "--many-faces",
            "--skip-audio",
            "--execution-provider", "cuda"
        ],
        capture_output=True,
        text=True
    )

    print(result.stdout)
    print(result.stderr)

    if os.path.exists(output) and os.path.getsize(output) > 500000:
        print("SUCCESS:", vid)
        done += 1
    else:
        print("FAILED:", vid)
        failed.append(vid)

print("\nDONE:", done)
print("FAILED:", failed)
```
